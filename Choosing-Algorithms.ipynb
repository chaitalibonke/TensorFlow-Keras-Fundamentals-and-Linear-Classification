```python
"""Neural networks
should usually be the last approach you look at due to the hardware and difficulty
associated with getting them performant."""
```

```python
"""Neural networks will have two or more layers where each can focus on and adjust the
weights of a particular feature to arrive at a better outcome than just one based on
activator functions.
Activator functions are simply a mathematical formula that is used to
determine whether a neuron should be active or dormant."""
```

```python
"""Perceptron – One of the most straightforward ANN architectures, the perceptron
is made up of a layer of inputs and a layer of Threshold Logic Unitnits (TLU).
TLUs are neurons that compute a weighed sum based on the inputs it receives. The
activator function in this case is called a step function. In training, if there is an
incorrect prediction, the weights for other inputs are adjusted to move to the correct
answer. Perceptrons aren't great for learning non-linear, complex relationships
• Multilayer Perceptron (MLP) – What's better than one layer? More than one
layer. An MLP is simply adding more layers to a perceptron base to discover more
complex relationships.
• Convolutional Neural Networks (CNNs) – Common in vision problems as its able
to look at areas of a video/picture and gain insight about a region and not look at
things pixel by pixel.
• Deep Neural Networks (DNN) – Simply a term to denote when there are multiple
layers of the network. This will mean that hidden layers are included in addition to
the input and output layers. In theory there is no limit to how many layers that you
can have. This is also where the term Deep Learning (DL) comes in. MLPs are an
examples of a DNN."""
```

```python
#Anomaly Detection
#Create a toy dataset
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager
from sklearn import svm
```

```python
X=0.2*np.random.randn(50,2)
X_train = np.r_[X+3]
```

```python
X = 0.2 * np.random.randn(20, 2)
X_test = np.r_[X + 3]
X_outliers = np.random.uniform(low=-5, high=5, size=(20, 2))
```

```python
#Training a one class SVM with scikit-learn
from sklearn import svm
one_svm = svm.OneClassSVM(nu=0.1, kernel="rbf", gamma=0.1)
one_svm.fit(X_train)
y_pred_train = one_svm.predict(X_train)
y_pred_test = one_svm.predict(X_test)
y_pred_outliers = one_svm.predict(X_outliers)
```

```python
#Now we'll use these predictions and stack them with the testing input to be able to plot
X_outliers = np.column_stack((X_outliers, y_pred_outliers))
df = pd.DataFrame(X_outliers, columns= ['x','y','anomaly_value'])
```

```python
valid = df.query('anomaly_value == 1')
anomaly = df.query('anomaly_value == -1')
```

```python
plt.title("One SVM")
plt.scatter(X_train[:, 0], X_train[:, 1], c="white",
edgecolors="k")
plt.scatter(X_test[:, 0], X_test[:, 1], c="blueviolet",
edgecolors="k")
plt.scatter(valid.iloc[:, 0], valid.iloc[:, 1], c="gold",
edgecolors="k", label="normal_test")
plt.scatter(anomaly.iloc[:, 0], anomaly.iloc[:, 1],
marker="x",color="red", label="anomaly")
plt.legend()
plt.show()
#All those X signs are prime areas that would be very
#important to check and make sure that they aren't fraudulent
```

```python
#Second type of common anomaly detection : Isolation Forests
#isolation forest creates a path down tree nodes to separate items into
#anomalies or not. It assumes that things that are anomalies
#will stand out and be easier to parse, and so if the tree is shorter and 
#the terminating node is reached quicker it assumes it is an anomaly
```

```python
#Isolation forests are another ensemble approach, one in which the power 
#of averages comes into play. Each tree created is referred to as an 
#Isolation Tree (iTree).
```

```python
from sklearn.ensemble import IsolationForest
isol_forest = IsolationForest(n_estimators=300, random_state=33)
isol_forest.fit(X_train)
```

```python
#Prediction using training data . Get outliers
y_pred_train = isol_forest.predict(X_train)
y_pred_test = isol_forest.predict(X_test)
y_pred_outliers = isol_forest.predict(X_outliers)
```

```python
full_test_outliers = np.column_stack((X_outliers, y_pred_outliers))
normal = full_test_outliers[(full_test_outliers[:,2] > 0)]
anomaly = full_test_outliers[(full_test_outliers[:,2] < 0)]
plt.scatter(X_train[:, 0], X_train[:, 1], c="white",edgecolors="k")
plt.scatter(X_test[:, 0], X_test[:, 1], c="blueviolet",edgecolors="k")
plt.scatter(normal[:, 0], normal[:, 1], c="gold",edgecolors="k", label="normal_test")
plt.scatter(anomaly[:, 0], anomaly[:, 1],marker="x",color="red", label="anomaly")
plt.legend()
plt.show()

#The resulting graph shows that there was one entity from the outlier group
#that the model identified as a member of the normal group, which seems to be
#a valid assumption.
```

```python
X, y = shap.datasets.boston()
X[:5]
```

```python
y[:5]
```

```python
d_param = {
    "learning_rate": 0.01
}

model = xgboost.train(params=d_param,
                      dtrain=xgboost.DMatrix(X, label=y), 
                      num_boost_round=100)
```

```python
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X)
```

```python
i = 0
shap.force_plot(explainer.expected_value, shap_values[i,:], X.iloc[i,:])
"""The visualization below shows the explanations for one prediction based on i-th data.

red: positive impacts on the prediction
blue: negative impacts on the prediction"""
```

```python
shap.summary_plot(shap_values, X, plot_type="violin")
```
